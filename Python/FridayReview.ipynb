### While Loops

While looks basically tell the computer to run a piece of code as long as a certain condition is true 

Think of it like: "Keep doing this code while this is true"


```python
count = 1

while count <= 5:
    print("Count is:", count)
    count = count + 1 
# updating the variable is very important so we dont run into infinite loops
```

```python
password = "hello"

while password != 'hello': # While password does not equal 'hello', say incorrect and promt to enter the password again
    print('Incorect password')
    password = input('Enter the password: ')
print('Access granted')
```

### For Loops

Similar to while loop in sense that it repeats code.

A for loop goes through items in a list or range, and runs code for each item.


### First way: Traverse through a list

```python
fruits = ['apple','orange','banana']
for x in fruits: # For each
    print('I like', x)
```

### Second way: Using range function

range(n) gives you a list of numbers 0 to n-1

```python
for num in range(5):
    print(num)
```

### Third way: Using range function but choosing where to start and stop
range(start, stop)

```python
for x in range(1,4):
    print(x)
```

### Forth way:
range(start, stop, step)

```python
for x in range (2, 11, 2):
    print(x)
```

### Boolean Arrays:
Boolean is True(1) or False (0)

```python
import numpy as np
cat_colors = np.array(['brown', 'black', 'black', 'grey', 'white', 'orange', 'black'])
cat_names = np.array(['nemo', 'lady', 'simba', 'tramp', 'snow', 'kitty', 'star'])
```

```python
black_color = cat_colors == 'black' #name of my boolean array; set condition on right side
black_color
```

```python
# use boolean array to pull out other data that correspons to True
black_cats_bol = cat_names[black_color] #original data[boolean array]
black_cats_bol
```

```python
#another example

food = np.array(['pizza', 'pizza', 'orange', 'chips', 'grape', 'cheese'])
cost = np.array([2,3,5,6,1,4])

#pull cost of all pizza
pizzas = food == 'pizza'
pizza_cost = cost[pizzas]
pizza_cost
```

```python
#one more example

price = np.array([23,45,100,34,12])
weight = np.array([13,14,15,16,17])

#pull price of everything that weights more than 15

heavy = weight > 15 #boolean_array = (condition)
price_heavy = price[heavy]
price_heavy
```

### Bar Plots:

used for when you have one numerical dataset and one categorical dataset

```python
#Sample data
categories = ['apples', 'bananas', 'cherries'] #categories always x axis
values = [10, 24, 18]  #numbers always y axis
colors = ['red', 'yellow', 'pink']

plt.bar(categories, values, color = colors) # making each bar a seperate color
plt.title('Fruit Sales')
plt.xlabel('Fruits')
plt.ylabel('Number Sold')
```

### Histograms:

Histograms look similiar to bar plots but are used to group NUMERIC data into bins (ranges) and show how many fall into tht range

```python
#example data of scores
scores = [85, 87, 70, 91, 88, 73, 74, 90, 95, 85, 78, 82, 88, 84, 79, 87, 89]

plt.hist(scores, bins = 5, color = 'darkseagreen', edgecolor = 'black')
plt.title('Distribution of Test Scores')
plt.xlabel('Score Ranges')
plt.ylabel('Number of Students (Frequecy)')
plt.show()
```

### Scatter plots:

shows relationships between two sets of numbers. Each point represents one data point. Similar to line plots but without the line.

```python
hours = [1, 2, 3, 4, 5, 6]
scores = [50, 55, 65, 70,75, 85]

#How many hours a student vs test score
plt.scatter(hours, scores, color = 'salmon') #(x,y)
plt.title('Hours Studied vs. Test Scores')
plt.xlabel('Hours studies')
plt.ylabel('Test Scores')
plt.grid(True)
```

### Subplots In Matplotlib:

Multiple plots in one figure.

```python
#Sample Data 

#1st Subplot
x_sub1 = [1,2,3,4]
y_sub1 = [2,4,6,8]

#2nd Subplot
x_sub2 = ['Apples', 'Bananas', 'Cherries']
y_sub2 = [6, 8, 10]

plt.subplot(1, 2, 1) # (rows, columns, which number plot you want to create)
plt.plot(x_sub1, y_sub1, colo = 'skyblue', marker='o', markerfacecolor = 'c')
plt.annotate("Peak", arrowprops = dict(facecolor = 'darkblue'), xy = (4,8), xytext = (3,8))
plt.title('1st Subplot')
plt.xlabel('x data')
plt.ylabel('y data')

colors = ['red', 'yellow', 'pink']
plt.subplot(1, 2, 2) # (rows, columns, and now i want to make the second plot)
plt.bar(x_sub2, y_sub2, color = colors)
plt.title('Fruitsb Sold')
plt.xlabel('Fruits')
plt.ylabel('Number Sold')

plt.tight_layout() # fixes layout
```
