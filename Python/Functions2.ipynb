Grocery Cost:
============
Return: cost_groceries

Variables: num_apples, num_oranges

```python
def grocery_cost(num_apples, num_oranges):
    price_apples = 3
    price_oranges = 2
    cost_apples = num_apples * price_apples
    cost_oranges = num_oranges * price_oranges
    total_cost = cost_apples + cost_oranges
    return total_cost
    
total=grocery_cost(2,4)
print(total)
```

_Celsius to Farenhiet_:
=======================

Return: farehheit

Variables: celsius

```python
def celcius_to_farenheit(celsius):
    temp_in_f = (celsius * 1.8) + 32
    return temp_in_f

temp_in_c = 20
c_to_f = celcius_to_farenheit(temp_in_c)
print(c_to_f)
```

_Farenheit to Celcius_:
=======================

Return: celsius

Variables: farenheit

```python
def farenheit_to_celcius(farenheit):
    temp_in_c = (farenheit - 32) * (5/9)
    return temp_in_c

temp_in_f = 68
f_to_c = farenheit_to_celcius(temp_in_f)
print(f_to_c)
```
