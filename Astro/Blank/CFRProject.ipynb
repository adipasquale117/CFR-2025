```python
import pandas as pd
import numpy as np
import h5py
import matplotlib.pyplot as plt
import matplotlib
```

```python
# Import spiral data
spiral_1 = pd.read_csv('../Data/CFR_project_data//belle/p_spiral_0.5.csv')
spiral_2 = pd.read_csv('../Data/CFR_project_data/belle/p_spiral_1.5.csv')
spiral_3 = pd.read_csv('../Data/CFR_project_data/belle/p_spiral_1.csv')
spiral_4 = pd.read_csv('../Data/CFR_project_data//belle/p_spiral_2.5.csv')
spiral_5 = pd.read_csv('../Data/CFR_project_data/belle/p_spiral_2.csv')
spiral_6 = pd.read_csv('../Data/CFR_project_data/belle/p_spiral_3.csv')
spiral_1
```

```python
# spiral variables
spiral1 = spiral_1['Probability is spiral (decimal)']
spiral2 = spiral_2['Probability is spiral (decimal)']
spiral3 = spiral_3['Probability is spiral (decimal)']
spiral4 = spiral_4['Probability is spiral (decimal)']
spiral5 = spiral_5['Probability is spiral (decimal)']
spiral6 = spiral_6['Probability is spiral (decimal)']

# turn these probabilities into a spiral fraction for each redshift
prob_spiral_1 = spiral1 >= .8
print(prob_spiral_1)

prob_spiral_2 = spiral2 >= .8
print(prob_spiral_2)

prob_spiral_3 = spiral3 >= .8
print(prob_spiral_3)

prob_spiral_4 = spiral4 >= .8
print(prob_spiral_4)

prob_spiral_5 = spiral5 >= .8
print(prob_spiral_5)

prob_spiral_6 = spiral6 >= .8
print(prob_spiral_6)
```

```python
# % of spiral galaxies variable
sum1 = round(sum(prob_spiral_1)/len(prob_spiral_1) * 100,2)
sum2 = round(sum(prob_spiral_2)/len(prob_spiral_2) * 100,2)
sum3 = round(sum(prob_spiral_3)/len(prob_spiral_3) * 100,2)
sum4 = round(sum(prob_spiral_4)/len(prob_spiral_4) * 100,2)
sum5 = round(sum(prob_spiral_5)/len(prob_spiral_5) * 100,2)
sum6 = round(sum(prob_spiral_6)/len(prob_spiral_6) * 100,2)

# print % of spiral galaxies variable
print('The percentage of spiral galaxies for redshift 0.5 is', sum1, '%')
print('The percentage of spiral galaxies for redshift 1.5 is', sum2, '%')
print('The percentage of spiral galaxies for redshift 1 is', sum3, '%')
print('The percentage of spiral galaxies for redshift 2.5 is', sum4, '%')
print('The percentage of spiral galaxies for redshift 2 is', sum5, '%')
print('The percentage of spiral galaxies for redshift 3 is', sum6, '%')
```

```python
# mean and standard deviation for redshift 0.5
print('The mean of spiral galaxies for redshift 0.5 is', round(np.mean(prob_spiral_1), 2))
print('The standard deviation of spiral galaxies for redshift 0.5 is', round(np.std(prob_spiral_1), 2))
print('')

# mean and standard deviation for redshift 1.5
print('The mean of spiral galaxies for redshift 1.5 is', round(np.mean(prob_spiral_2), 2))
print('The standard deviation of spiral galaxies for redshift 1.5 is', round(np.std(prob_spiral_2), 2))
print('')

# mean and standard deviation for redshift 1
print('The mean of spiral galaxies for redshift 1 is', round(np.mean(prob_spiral_3), 2))
print('The standard deviation of spiral galaxies for redshift 1 is', round(np.std(prob_spiral_3), 2))
print('')

# mean and standard deviation for redshift 2.5
print('The mean of spiral galaxies for redshift 2.5 is', round(np.mean(prob_spiral_4), 2))
print('The standard deviation of spiral galaxies for redshift 2.5 is', round(np.std(prob_spiral_4), 2))
print('')

# mean and standard deviation for redshift 2
print('The mean of spiral galaxies for redshift 2 is', round(np.mean(prob_spiral_5), 2))
print('The standard deviation of spiral galaxies for redshift 2 is', round(np.std(prob_spiral_5), 2))
print('')

# mean and standard deviation for redshift 3
print('The mean of spiral galaxies for redshift 3 is', round(np.mean(prob_spiral_6), 2))
print('The standard deviation of spiral galaxies for redshift 3 is', round(np.std(prob_spiral_6), 2))
print('')
```

```python
# import data from Astrobytes article 
data = pd.read_csv('..//Data/CFR_project_data//belle/data.csv')
data
```

```python
# % of spiral galaxies variable
x2 = data.values[:, 0]
y2 = data.values[:, 1] * 100
```

```python
# Make variables for simulated data
x1 = [8.587, 5.878, 4.293, 3.285, 2.685, 2.145]
y1 = [sum1,sum3,sum2,sum5,sum4,sum6]

# Make variables for JWST data
x2 = [7.132, 5.115, 3.744, 2.981, 2.384, 1.944] 
y2 = data.values[:, 1] * 100

# Make Scatter Plot w/ Simulated and JWST data
plt.scatter(x1, y1, color = 'cornflowerblue', label = 'Simulated Universe data')
plt.plot(x1, y1, color = 'cornflowerblue', linestyle = '--', alpha = 0.7)

plt.scatter(x2, y2, color = 'k', label = 'Our Universe (JWST data)')
plt.plot(x2, y2, color = 'k', linestyle = '-', alpha = 0.7)

# Label graph
plt.xlabel('Age of Universe [Gyrs]')
plt.ylabel('Spiral Fraction')
plt.title('Spiral Fraction Vs. Age of Universe [Gyrs]')

plt.legend()
plt.xlim(0, 9)

plt.show()
```
